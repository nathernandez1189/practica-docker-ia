# 2. TensorFlow Sample

```python
# TensorFlow version
import tensorflow as tf
import datetime
import numpy as np
tf.keras.utils.set_random_seed(42)

print(tf.__version__)
```

```python
# Load the TensorBoard notebook extension
%load_ext tensorboard
```

```python
#Clear any logs from previous runs
# Se conservan los registros de las ejecuciones anteriores.
```

```python
mnist = tf.keras.datasets.mnist

(x_train, y_train),(x_test, y_test) = mnist.load_data()
x_train, x_test = x_train[:2000].astype("float32") / 255.0, x_test[:500].astype("float32") / 255.0
y_train, y_test = y_train[:2000], y_test[:500]

def create_model():
  return tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(512, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation='softmax')
  ])
```

## Using TensorBoard with Keras Model.fit()¶

```python
model = create_model()
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

log_dir="logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorboard_callback = tf.keras.callbacks.TensorBoard(log_dir=log_dir, histogram_freq=1)

for epoch in range(3):
  for start in range(0, len(x_train), 64):
    model.train_on_batch(x_train[start:start+64], y_train[start:start+64])
predicted = np.argmax(model(x_test, training=False).numpy(), axis=1)
accuracy = float(np.mean(predicted == y_test))
print(f"MNIST: exactitud sobre 500 imágenes = {accuracy:.4f}")
assert accuracy > 0.70
```
